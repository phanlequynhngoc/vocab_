# 1. Sau mỗi lần sửa code
Chạy:
```bash
git status
```
Nếu thấy file thay đổi: 
```bash
git add .
git commit -m "mô tả thay đổi"
git push
```
Ví dụ:
```bash
git commit -m "Add pronunciation in review page"
```
# 2. Kiểm tra Vercel trước khi test

```bash
Vercel
→ Deployments
→ Latest deployment
```

> Production
